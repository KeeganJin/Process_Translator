```python
from Utils.visual_backend import (
    get_petri_net_structure,
    extract_subnet_visual_elements,
    assign_colors_to_patterns,
    generate_petri_net_dot
)
pnml_path = "./uploads/ID_1.pnml"
# Step 1: Get net data
net_data = get_petri_net_structure(pnml_path)

pattern_mapping = [
    {
        "pattern_name": "pattern_basic_xor_1",
        "edge_mapping": [
            {"a": "Back-order Part", "b": "Reserve Part"}
        ]
    },
    {
        "pattern_name": "pattern_PETRINET_1_1",
        "edge_mapping": [
            {"a": "Check Part Quality", "b": "Back-order Part", "c": "Reserve Part", "d": "Select Unchecked Part"}
        ]
    }
]


# Step 2: Get pattern subnets (from your detection result)
pattern_subnets = extract_subnet_visual_elements(pnml_path, pattern_mapping)
print(pattern_subnets)
# Step 3: Assign colors (optional but recommended for clarity)
color_map = assign_colors_to_patterns([p["pattern_name"] for p in pattern_subnets])
for p in pattern_subnets:
    p["color"] = color_map.get(p["pattern_name"], "#888")

# Step 4: Generate DOT
dot = generate_petri_net_dot(net_data, pattern_subnets)

# Step 5: Render SVG/PNG
import graphviz
graph = graphviz.Source(dot)
graph.render('annotated_petri_net', format='svg', view=True)
```
